# Draw outlines of the `mast-aladin` viewport in a `jdaviz` viewer, and vice versa

In this tutorial, we will:
* create instances of `MastAladin` and `jdaviz`
* draw outlines of each viewport in the other
* toggle outlines on/off, and change their styles

### Imports

````python
import jdaviz as jd
from mast_aladin import MastAladin, AppSidecar, ViewportOutline
````

### Initialize apps

Initialize `MastAladin`, centered on the globular cluster M4, and load an image centered on M4 into `jdaviz`. Display both in sidecars.

````python
mast_aladin = MastAladin(
    target='M4',
    fov=0.3,
)

jd.load('mast:JWST/products/jw01979002001_02101_00001_nrcb4_cal.fits', format = "Image")
jd.plugins['Orientation'].align_by = 'WCS'

# open sidecars for both mast-aladin and jdaviz:
mast_aladin, viz = AppSidecar.open(mast_aladin, jd.gca(), anchor=["split-right", "split-bottom"])
````

### Render viewport outlines between the apps

Draw the viewport outline of `jdaviz` default viewer in `mast-aladin` and vice versa.

Note: for now, the cell below must be run *after* the above have completed rendering outputs, otherwise an error will be raised. If the error is raised but both widgets have since completed rendering, rerun the cell below.

````python
outlines = ViewportOutline.for_current_apps()
````

Note: the above call is equivalent to ```outlines = ViewportOutline(viz.default_viewer, aladin)```.

When working with multiple viewers, be sure so specify which apps you want to share viewport outlines between.

At the initial zoom levels, `mast-aladin` shows the full extent of the globular cluster, and `jdaviz` shows a single JWST NIRCam image of a small subset of stars in the cluster. If you zoom in on `mast-aladin` towards the red outline of the `jdaviz` viewport, you will see the red outline of the `mast-aladin` viewport in `jdaviz`. If you zoom in on `jdaviz` onto a particular star, you will see the red outline of the `jdaviz` viewport in `mast-aladin`.

The outlines in each app can be shown or cleared by setting `aladin_outline_in_jdaviz` and `jdaviz_outline_in_aladin`:

````python
# stop rendering the outlines
outlines.aladin_outline_in_jdaviz = False
outlines.jdaviz_outline_in_aladin = False
````

````python
# turn back on rendering the outlines
outlines.aladin_outline_in_jdaviz = True
outlines.jdaviz_outline_in_aladin = True
````

The outline styling can be updated using the following commands:

````python
# change the outline color
outlines.outline_color = '#00ff00'
````

````python
# change the outline width
outlines.outline_width = 6
````
