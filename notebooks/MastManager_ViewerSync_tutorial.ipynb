# `MastManager` `ViewerSync` tutorial

```python
from mast_aladin import MastAladin, MastManager
from sidecar import Sidecar
import jdaviz as jd
```

### Setup

Create an instance of the MastManager class and then `jdaviz` and `MastAladin` instances to manage.

```python
# create MastManager instance
mm = MastManager()
```

```python
# create jdaviz instance
sc = Sidecar(title="jdaviz Main", anchor="split-right")

with sc:
    jd.show()
jd.load("mast:JWST/product/jw02727-o002_t062_nircam_clear-f090w_i2d.fits", format="Image", cache=True)
```

<div class="alert alert-block alert-info">
    Note: It is important to load the image(s) you're interested in before registering the jdaviz app.
</div>

```python
# register jdaviz instance
mm.register_app(jd.gca(), "viz")
```

Create two instances of `MastAladin`

```python
# create and register first MastAladin instance
mast_aladin_color = MastAladin(target="Cartwheel Galaxy")

sc = Sidecar(title="mast-aladin-color", anchor="split-bottom", ref=sc)
with sc:
    display(mast_aladin_color)

mm.register_app(mast_aladin_color, "mast-aladin-color")
```

```python
# create and register second MastAladin instance
mast_aladin_red = MastAladin(target="Cartwheel Galaxy", survey="P/DSS2/red")

sc = Sidecar(title="mast-aladin-red", anchor="split-right", ref=sc)
with sc:
    display(mast_aladin_red)

mm.register_app(mast_aladin_red, "mast-aladin-red")
```

### Viewer Sync

```python
mm.display_plugin('viewer_sync')
```

### AIDA API
The AIDA API provides the ability to set the viewport aspects of compatible viewers (including `jdaviz` and `MastAladin`. Experiment with setting different aspect (like center, fov, and rotation) using the cells below to see how the ViwerSync keeps the viewports aligned!

```python
jd.viewers['Image'].set_viewport(fov = 5)
```

```python
mast_aladin_color.set_viewport(fov = 0.01)
```
