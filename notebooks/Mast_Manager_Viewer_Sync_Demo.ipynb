# Mast Manager Viewer Sync Demo

```python
from mast_aladin import MastAladin, MastManager
from sidecar import Sidecar
import jdaviz as jd
```

### Setup
Create an instance of the MastManager class 

```python
mm = MastManager()
```

**`Jdaviz`**

For this example we will create an instance of `jdaviz`.

```python
jdaviz = jd.App()
jdaviz.load("mast:JWST/product/jw02727-o002_t062_nircam_clear-f090w_i2d.fits", format="Image", cache=True)
```

Note: It is important to load the image(s) you're interested in before registering the jdaviz app

**`MastAladin`**

For this example we will create two instances of `MastAladin`

```python
mast_aladin_color = MastAladin(
    target="Cartwheel Galaxy",
    survey='https://stpubdata.s3.us-east-1.amazonaws.com/mast/skybackgrounds/DSSColor'
)
mast_aladin_red = MastAladin(
    target="Cartwheel Galaxy",
    survey='https://stpubdata.s3.us-east-1.amazonaws.com/mast/skybackgrounds/DSSColor'
)
```

### Display the created widgets 
we will utilize the built in `display_apps` method of `MastManager` to show the created widgets

```python
mm.display_apps()
```

### Viewer Sync Plugin

```python
mm.display_plugin('viewer_sync')
```

### AIDA API
The AIDA API provide the ability to sync the viewport between two AIDA supported viewers. 

```python
mast_aladin_color.set_viewport(fov=1)
```
