## Imports

```python
from astroquery.mast import Observations
from astropy.coordinates import SkyCoord
from mast_aladin import MastAladin
```

## Step 1: Creating a `MastAladin` Sky Viewer

To begin, we'll create a sky viewer using `MastAladin`. This viewer is an interactive widget that allows you to explore sky regions from your Jupyter notebook.

You can initialize with custom values for:
- The **target** (center of the view) — using object names (e.g., `"M31"`, `"Andromeda Galaxy"`), or coordinates (`SkyCoord`)
- The **field of view** (`fov`) — how much of the sky you see, in degrees
- The **viewer height** — how tall the widget appears on your screen

 **Try it yourself**: After running the cell below, try changing the `target`, `fov`, or `height` values to see how the viewer appears!

```python
mast_aladin = MastAladin(
    target="M31",   # accepts string or SkyCoord
    fov=5,          # field of view in degrees
    height=500      # widget height in pixels
)
mast_aladin
```

## Step 2: Query MAST for Data around a Target

Now that we have a viewer showing a region of the sky, let's find **real science data** from [MAST (Mikulski Archive for Space Telescopes)](https://archive.stsci.edu/).

By leveraging the `Observations` class from [`astroquery.mast`](https://astroquery.readthedocs.io/en/latest/mast/mast.html), we will:

- Specify a sky target (e.g., Andromeda Galaxy – “M31”)
- Retrieve available observations within a search radius around that target
- Display some of the table of observations


```python
# Use `SkyCoord.from_name("M31")` to automatically resolve the name "M31" into sky coordinates (RA/Dec).
m31 = SkyCoord.from_name("M31")

# Search within a 0.1° radius for any MAST observations
results = Observations.query_region(m31, radius="0.1 deg")

# Display the first few rows of the result table
results[:5]
```

## Step 3: Filter and Display Observations in `MastAladin`

Now that we have a table of MAST observations, let’s filter them down to something specific.

In this example, we’ll select only the rows with the observation title **"PS1 3PI Survey"**, but you can filter the table however you like to fit your own workflow (e.g. by instrument, date, or collection).

Once you’ve filtered the table, simply pass it into `add_table()` to display the selected observations as overlays.

```python
# Filter the results table by observation title
filtered = results[results["obs_title"] == "PS1 3PI Survey"]

# Add the filtered observations to the viewer
mast_aladin.add_table(filtered)
```

## Step 4: Save and Reuse Viewport Settings

Once you’ve adjusted the viewer to focus on a region of interest, you might want to **save** that position and **return to it later** — or even share it with someone else.

You can do this using the `get_viewport()` method. It returns a dictionary with:

- `center`: where the viewer is centered (a `SkyCoord`)
- `fov`: the field of view (length of the shorter viewport axis.)
- `rotation` : angle of the view center to north pole angle in degrees.
- `projection`: the projection of the viewer.
- `image_label`: A str representing the label of the image, always `None` for mast-aladin.

Let’s try it out:

```python
# Get the current viewport
viewport = mast_aladin.get_viewport()
viewport
```

Now, let’s say you want to return to this same view later. You can pass the center value back into `set_viewport()`:

```python
# Set the viewport using the saved center from earlier
mast_aladin.set_viewport(center=viewport["center"])
```

## Step 5: Display `MastAladin` in a Sidecar Panel

To keep the notebook clean and focused, you can show the `MastAladin` viewer in a separate side panel using the `.show()` method.

This keeps the viewer visible while you continue to scroll and interact with other content, which is especially helpful when you're working with large data tables or running long cells.

Let’s display a `MastAladin` viewer in the default split-right sidecar:

```python
mast_aladin = MastAladin(
    target="Cartwheel Galaxy",
    fov=0.2
)
mast_aladin.show()
```

You can customize the position of the viewer sidecar by setting the anchor option. Available options include:

    'sidecar:split-right' (default), 'sidecar:split-left', 'sidecar:split-top', 
    'sidecar:split-bottom', 'sidecar:tab-before', 'sidecar:tab-after', 'sidecar:right'

For example, here’s how to dock the viewer at the bottom of your notebook:

```python
mast_aladin = MastAladin(
    target="M1",
    fov=3
)
mast_aladin.show("sidecar:split-bottom")
```
