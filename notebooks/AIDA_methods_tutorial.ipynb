# Astro Image Display API (AIDA) methods tutorial

In this tutorial, we demonstrate how to use the [Astro Image Display API (AIDA)](https://github.com/astropy/astro-image-display-api/) methods available in `mast-aladin` to allow for easy updating of viewports. These include:
- `get_viewport()`
- `set_viewport()`

### Imports

```python
from mast_aladin import MastAladin
from astropy.coordinates import SkyCoord, Angle
```

### Initialize `mast-aladin`

```python
mast_aladin = MastAladin(
    target = "269.999666,65.9000833",
    fov = 1.5,
    height = 800,
)
mast_aladin.show()
```

```python
# Add a footprint for easy visual reference of functionality
# mast-aladin can load a footprint from an STC-S string like this:
stcs_string = "POLYGON ICRS 269.986903302 65.984279763 269.986696947 66.107243454 269.676617336 66.107210320 269.679951241 65.984356587"
polygon = mast_aladin.add_graphic_overlay_from_stcs(stcs_string)
```

### Using `set_viewport` and `get_viewport`

We can use the AIDA methods to set the parameters of the viewport in `mast-aladin`. The AIDA aspects available to be set in `mast-aladin` include: center, fov, rotation, and projection. 

Note: though `image label` is an AIDA aspect, it has no relevance for a HiPS browser like `mast-aladin`. This aspect is relevant for other widgets like `jdaviz`.

```python
# Recenter on corner of polygon overlay
mast_aladin.set_viewport(
    center = SkyCoord(269.986903302, 65.984279763, unit="deg"),
    fov = Angle(0, unit="deg"),
)
```

We can then use `get_viewport` to save the state for later use.

Note: `mast-aladin` has no concept of pixels. Trying to run `mast_aladin.get_viewport(sky_or_pixel="pixel")` will raise an error.

```python
viewport_state = mast_aladin.get_viewport()
viewport_state
```

Now, let's consider the case you want to focus on a different region. To do so, we can run the cell below:

```python
# Center the viewer on RA=0 deg, Dec=0 deg, set rotation to 180 deg, and change the projection to "TAN"
mast_aladin.set_viewport(
    center=SkyCoord(0,0, unit="deg"),
    rotation=Angle(180, unit="deg"),
    projection="TAN",
)
```

After looking at the region we just moved to, we can return to the earlier viewport state centered on the corner of the polygon overlay by running the following cell:

```python
mast_aladin.set_viewport(
    center = viewport_state["center"],
    rotation = viewport_state["rotation"],
    projection = viewport_state["projection"],
)
```
